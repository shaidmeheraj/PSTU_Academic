# Genetic / Evolutionary Algorithm – Concept (সহজভাবে)

Idea:

Nature inspired → Darwin’s “Survival of the fittest”।

Problem solution = population of candidate solutions।

প্রতিটি iteration-এ (generation):

Selection → best candidates বাছাই করা।

Crossover (Recombination) → 2 candidates মিলে new child তৈরি।

Mutation → ছোট random change introduce করা → diversity বজায় রাখা।

Repeat until solution found বা max generations।

Advantages:

Randomness + selection → local maxima এ আটকে থাকার সমস্যা কমে।

অনেক combinatorial / NP-hard problems solve করতে পারি।

Analogy:

Population of boards (like creatures), যারা “strongest” হয় → next generation এ survive করে, weak boards die out।

```python
import random

N = 8  # Number of queens
POP_SIZE = 200  # Bigger population = more diversity
MUTATION_RATE = 0.2
MAX_GENERATIONS = 5000

# Heuristic: number of attacking pairs (lower is better)
def fitness(state):
    attacks = 0
    for i in range(N):
        for j in range(i+1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                attacks += 1
    return -(attacks)  # we maximize negative attacks

# Create random state
def random_state():
    return [random.randint(0, N-1) for _ in range(N)]

# Roulette wheel selection
def select(population):
    fitnesses = [fitness(ind) for ind in population]
    min_fit = min(fitnesses)
    # Shift fitness to positive values
    fitnesses = [f - min_fit + 1 for f in fitnesses]
    total = sum(fitnesses)
    pick = random.uniform(0, total)
    current = 0
    for ind, fit in zip(population, fitnesses):
        current += fit
        if current >= pick:
            return ind
    return population[-1]

# Crossover: single-point
def crossover(parent1, parent2):
    point = random.randint(1, N-2)
    return parent1[:point] + parent2[point:]

# Mutation: randomly change one queen
def mutate(state):
    if random.random() < MUTATION_RATE:
        col = random.randint(0, N-1)
        row = random.randint(0, N-1)
        state[col] = row
    return state

# Genetic Algorithm
def GENETIC_ALGORITHM():
    population = [random_state() for _ in range(POP_SIZE)]

    for generation in range(MAX_GENERATIONS):
        # Check if solution exists
        for state in population:
            if fitness(state) == 0:
                return state, generation
        
        new_population = []
        # Elitism: keep the best one
        best = max(population, key=fitness)
        new_population.append(best)

        while len(new_population) < POP_SIZE:
            parent1 = select(population)
            parent2 = select(population)
            child = crossover(parent1, parent2)
            child = mutate(child)
            new_population.append(child)

        population = new_population

    return None, MAX_GENERATIONS

# Run the algorithm
solution, gen = GENETIC_ALGORITHM()
if solution:
    print("Solution found in generation", gen)
    print("Solution:", solution)
else:
    print("No solution found")
```
