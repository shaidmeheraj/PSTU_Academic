# Local Beam Search – Concept (সহজভাবে)

Idea:

Hill Climbing একটাই state ধরে neighbor explore করে।

Local Beam Search k states একসাথে ধরে রাখে → multiple “paths” একসাথে explore করে।

প্রতিটি iteration-এ:

সব k states-এর neighbor তৈরি করা হয়।

সব neighbors-এর মধ্যে best k states নির্বাচন করা হয়।

Repeat until solution found or stop condition।

Advantages:

একাধিক paths একসাথে explore করা → local maxima এ আটকে থাকার সম্ভাবনা কমে।

Random restart-র দরকার নেই যদি k ভালো হয়।

Disadvantages:

বেশি memory & computation লাগে, কারণ k states manage করতে হয়।

Analogy:

ধরো তুমি ৫ জন climber নিয়ে পাহাড়ে উঠছো।

প্রতিটি climber নিজ নিজ দিক explore করে → প্রতিটি step-এ সবচেয়ে promising ৫টি position রাখো → এইভাবে hill-climbing এর অনেক local maxima এ আটকে যাওয়ার সমস্যা কমে।

```python
import random

N = 8  # Number of queens
K = 3  # Number of states to keep (beam width)

# Heuristic: number of attacking pairs
def VALUE(state):
    attacks = 0
    for i in range(N):
        for j in range(i+1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                attacks += 1
    return -attacks  # maximize negative attacks

# Generate all successors of a state
def get_successors(state):
    successors = []
    for col in range(N):
        for row in range(N):
            if row != state[col]:
                neighbor = state.copy()
                neighbor[col] = row
                successors.append(neighbor)
    return successors

# Local Beam Search Algorithm
def LOCAL_BEAM_SEARCH(K):
    # Initialize K random states
    states = [[random.randint(0, N-1) for _ in range(N)] for _ in range(K)]

    while True:
        # Check if any state is solution
        for state in states:
            if VALUE(state) == 0:
                return state

        # Generate all successors of all states
        all_successors = []
        for state in states:
            all_successors.extend(get_successors(state))

        # Select best K successors
        states = sorted(all_successors, key=VALUE, reverse=True)[:K]

# Run the algorithm
solution = LOCAL_BEAM_SEARCH(K)
print("Solution:", solution)
print("Heuristic value (attacking pairs):", -VALUE(solution))
```
