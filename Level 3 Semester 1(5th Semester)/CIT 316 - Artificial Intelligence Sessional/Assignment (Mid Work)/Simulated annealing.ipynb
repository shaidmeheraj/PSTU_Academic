# 4.1.2 : Simulated Annealing

```python
import random
import math

N = 8  # Number of queens

# Heuristic: number of attacking pairs (we want to minimize this)
def VALUE(state):
    attacks = 0
    for i in range(N):
        for j in range(i+1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                attacks += 1
    return attacks  # Smaller is better

# Generate a random successor by moving one queen in its column
def random_successor(state):
    neighbor = state.copy()
    col = random.randint(0, N-1)
    row = random.randint(0, N-1)
    neighbor[col] = row
    return neighbor

# Cooling schedule function
def schedule(t):
    return max(0.01, 100 * (0.95 ** t))  # T decreases over time

# Simulated Annealing Algorithm
def SIMULATED_ANNEALING(problem, max_steps=1000):
    current = problem['INITIAL']
    
    for t in range(1, max_steps + 1):
        T = schedule(t)
        if T == 0:
            return current
        
        next_state = random_successor(current)
        delta_E = VALUE(current) - VALUE(next_state)
        
        if delta_E > 0:
            current = next_state  # Better state
        else:
            probability = math.exp(delta_E / T)
            if random.random() < probability:
                current = next_state  # Accept worse state with probability
        
        if VALUE(current) == 0:  # Solution found
            return current
    
    return current  # Return best found state

# Random initial state
initial_state = [random.randint(0, N-1) for _ in range(N)]
problem = {'INITIAL': initial_state}

solution = SIMULATED_ANNEALING(problem)
print("Initial state:", initial_state)
print("Final state:", solution)
print("Heuristic value (attacking pairs):", VALUE(solution))
```

| Feature                                  | Hill Climbing                                                          | Simulated Annealing                                                             |
| ---------------------------------------- | ---------------------------------------------------------------------- | ------------------------------------------------------------------------------- |
| **Neighbor Selection**                   | Always selects **best neighbor** → local best pick                     | Chooses **random neighbor**, can sometimes accept worse neighbor                |
| **Escape from Local Maxima**             | ❌ Cannot escape → stuck easily                                         | ✅ Can escape local maxima with **probabilistic bad moves**                      |
| **Probability of Accepting Worse Moves** | 0 (never)                                                              | >0, depends on **temperature T** → `exp(ΔE / T)`                                |
| **Temperature / Cooling**                | Not used                                                               | Uses **temperature (T)** and **cooling schedule** to gradually reduce bad moves |
| **Speed**                                | Fast & simple                                                          | Slower than HC, depends on steps and cooling schedule                           |
| **Likelihood of Finding Global Optimum** | Low → may get stuck in **local maxima**                                | High → can reach **global maximum** with proper schedule                        |
| **Analogy**                              | Blindfolded climber always moves uphill → may get stuck on small hills | Blindfolded climber sometimes steps down → can reach highest mountain           |
