```python
def f(x):
    # Example function: f(x) = x^3 - x - 2
    return x**3 - x - 2

def false_position(a, b, tol=1e-6, max_iter=100):
    if f(a) * f(b) > 0:
        print("The function must have opposite signs at a and b")
        return None

    print("Iter\t a\t\t b\t\t c\t\t f(c)")
    for i in range(max_iter):
        # Apply False Position Formula
        c = (a * f(b) - b * f(a)) / (f(b) - f(a))
        fc = f(c)

        print(f"{i+1}\t{a:.6f}\t{b:.6f}\t{c:.6f}\t{fc:.6f}")

        if abs(fc) < tol:
            print(f"\nRoot found: {c:.6f} after {i+1} iterations")
            return c

        # Update interval
        if f(a) * fc < 0:
            b = c
        else:
            a = c

    print("\nMax iterations reached")
    return c

# Example usage
false_position(1, 2)
```
