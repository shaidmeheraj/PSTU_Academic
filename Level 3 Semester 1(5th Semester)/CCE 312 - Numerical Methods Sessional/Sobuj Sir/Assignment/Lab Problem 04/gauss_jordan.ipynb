```python
# Gauss-Jordan Method

def gauss_jordan(a, b):
    n = len(b)

    # Form augmented matrix
    for i in range(n):
        a[i].append(b[i])

    print("Initial Augmented Matrix:")
    for row in a:
        print(row)
    print()

    # Applying Gauss-Jordan Elimination
    for i in range(n):
        # Make pivot = 1
        pivot = a[i][i]
        if pivot == 0:
            raise ValueError("Zero pivot encountered!")
        for j in range(i, n + 1):
            a[i][j] /= pivot

        # Make all other rows have 0 in column i
        for k in range(n):
            if k != i:
                factor = a[k][i]
                for j in range(i, n + 1):
                    a[k][j] -= factor * a[i][j]

        print(f"After eliminating column {i+1}:")
        for row in a:
            print(row)
        print()

    # Extract solution
    solution = [a[i][n] for i in range(n)]
    return solution


# Example:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

A = [
    [2.0, 1.0, -1.0],
    [-3.0, -1.0, 2.0],
    [-2.0, 1.0, 2.0]
]
B = [8.0, -11.0, -3.0]

solution = gauss_jordan(A, B)

print("Final Solution:")
for i, val in enumerate(solution):
    print(f"x{i+1} = {val}")
```
