```python
import numpy as np
```

```python
# Function to integrate
def f(x):
    return x**2

# Trapezoidal Rule
def trapezoidal(a, b, n):
    h = (b - a) / n
    result = f(a) + f(b)
    
    for i in range(1, n):
        x = a + i * h
        result += 2 * f(x)
    
    result *= (h / 2)
    return result
```

```python
# Example: ∫0^2 x^2 dx
a = 0
b = 2
n = 4   # number of subintervalsz
```

```python
approx = trapezoidal(a, b, n)
print(f"Approximate value using Trapezoidal Rule: {approx:.6f}")
```

```python
# Exact value for comparison
exact = (b**3 - a**3) / 3
print(f"Exact value: {exact:.6f}")
print(f"Error: {abs(exact - approx):.6f}")
```
