```python
#2x - y + 3z = 9
#x + z = 4
#3x + 2y - 2z = -10

import numpy as np

# Coefficient matrix A
A = np.array([[2, -1, 3],
              [1, 0, 1],
              [3, 2, -2]])

# Constant matrix B
B = np.array([9, 4, -10])

# Determinant of A
det_A = np.linalg.det(A)

if det_A == 0:
    print("No unique solution exists (det(A) = 0)")
else:
    # Replace columns for Cramer's Rule
    A_x = A.copy()
    A_x[:, 0] = B

    A_y = A.copy()
    A_y[:, 1] = B

    A_z = A.copy()
    A_z[:, 2] = B

    # Compute determinants
    det_x = np.linalg.det(A_x)
    det_y = np.linalg.det(A_y)
    det_z = np.linalg.det(A_z)

    # Apply Cramer's rule
    x = det_x / det_A
    y = det_y / det_A
    z = det_z / det_A

    print(f"Solution: x = {x:.2f}, y = {y:.2f}, z = {z:.2f}")
```
